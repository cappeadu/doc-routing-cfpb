```python
import pandas as pd
from sklearn.model_selection import train_test_split

val = pd.read_csv("../datasets/val.csv")

val_sample, _ = train_test_split(
    val,
    test_size=val.shape[0] - 100,
    stratify=val["labels"],
    random_state=42,
)

val_sample.to_csv("../dataset/val_sample.csv", index=False)

print(val_sample.shape)
print(val_sample["labels"].value_counts())
```
